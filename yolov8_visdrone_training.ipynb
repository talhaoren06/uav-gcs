```python
from ultralytics import YOLO

# Load the YOLOv8 Nano model from scratch
model = YOLO("yolov8n.pt")

# Start the training process with defined parameters
results = model.train(
    data="VisDrone.yaml",
    epochs=100,            # Maximum number of training iterations
    patience=15,           # Early stopping mechanism to prevent overfitting
    imgsz=640,             # Standard input resolution for UAV cameras
    batch=16,              # Batch size optimized for T4 GPU memory
    device=0,              # Execute training on GPU (Device 0)
    save=True,             # Automatically save the best model weights
    project="UAV_Project", # Main project directory name for saving results
    name="VisDrone_Model"  # Subdirectory name for this specific training run
)
```
